```python
import pandas as pd
file_name = "carbon_emission_dataset_with_Industry.csv"
df = pd.read_csv(file_name)

print("\n أوّل 5 أسطر من البيانات")
display(df.head())

print("\nإحصاءات عامة ومبدئيّة عن الجدول للأعمدة الرقميّة")
display(df.describe())
```

```python
print("فحص القيم الفارغة في كل عمود:")
print(df.isnull().sum())

df_cleaned = df.dropna()

print("\n تم تنظيف البيانات بنجاح!")
print("عدد الأسطر المتبقّية للتحليل: {len(df_cleaned)}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# الأعمدة المطلوبة للتحليل
target_columns = [
    'Total_Energy_Consumption_kWh',
    'Renewable_Energy_Consumption_kWh',
    'NonRenewable_Energy_Consumption_kWh',
    'Production_Output_Units',
    'Supply_Chain_Transport_km'
]

# 1. حساب الإحصاءات الأساسية لكل عمود
print(" الإحصاءات الأساسية للمتغيرات الخمسة ")
for col in target_columns:
    print(f"\n[ المتغير: {col} ]")
    print(f"  - المتوسط الحسابي (Mean): {df_cleaned[col].mean():.2f}")
    print(f"  - الوسيط (Median): {df_cleaned[col].median():.2f}")
    print(f"  - أعلى قيمة (Max): {df_cleaned[col].max():.2f}")
    print(f"  - أقل قيمة (Min): {df_cleaned[col].min():.2f}")

# 2. رسم توزيع البيانات (Distribution Plots)
plt.figure(figsize=(15, 12))
for i, col in enumerate(target_columns, 1):
    plt.subplot(3, 2, i)
    sns.histplot(df_cleaned[col], kde=True, color='teal')
    plt.title(f'Distribution of {col}', fontsize=12)
    plt.xlabel('')
    plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# 1. رسم مخطط تكرار القطاعات الصناعية (أعلى 10 قطاعات)
plt.figure(figsize=(12, 5))
sns.countplot(data=df_cleaned, y='Industry_Sectors',
              order=df_cleaned['Industry_Sectors'].value_counts().index[:10], palette='muted')
plt.title('Top 10 Industry Sectors Frequency', fontsize=14)
plt.xlabel('Count')
plt.ylabel('Industry Sector')
plt.show()

# 2. خريطة الترابط الحرارية (Correlation Heatmap) للمتغيرات الرقمية
plt.figure(figsize=(12, 8))
numeric_cols = df_cleaned.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_cols.corr(), annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Variables', fontsize=14)
plt.show()

# 3. مخطط الصندوق للانبعاثات الكربونية حسب القطاع الصناعي (أعلى 5 قطاعات للوضوح)
plt.figure(figsize=(12, 6))
top_5_sectors = df_cleaned['Industry_Sectors'].value_counts().index[:5]
df_top_5 = df_cleaned[df_cleaned['Industry_Sectors'].isin(top_5_sectors)]

sns.boxplot(data=df_top_5, x='Industry_Sectors', y='Carbon_Emission_tCO2e_TARGET', palette='Set2')
plt.title('Carbon Emissions Grouped by Top 5 Industry Sectors', fontsize=14)
plt.xticks(rotation=15)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from statsmodels.tsa.seasonal import seasonal_decompose

# 1. تحويل عمود الوقت إلى صيغة زمنية يفهمها بايثون
df_cleaned['Date'] = pd.to_datetime(df_cleaned['Date'])
df_cleaned = df_cleaned.sort_values('Date') # ترتيب البيانات من الأقدم للأحدث

# 2. رسم الانبعاثات مع مرور الوقت
plt.figure(figsize=(15, 5))
plt.plot(df_cleaned['Date'], df_cleaned['Carbon_Emission_tCO2e_TARGET'], label='Carbon Emissions', color='purple')
plt.title('Carbon Emissions Over Time')
plt.xlabel('Date')
plt.ylabel('Emissions')
plt.legend()
plt.show()

if 'Temperature' in df_cleaned.columns:
    plt.figure(figsize=(15, 5))
    plt.plot(df_cleaned['Date'], df_cleaned['Temperature'], label='Temperature', color='orange')
    plt.title('Temperature Trend Over Time')
    plt.xlabel('Date')
    plt.ylabel('Temperature')
    plt.legend()
    plt.show()

# 3. الإحصاءات المتحركة (Rolling Statistics: المتوسط المتحرك والتباين المتحرك)
rolling_mean = df_cleaned['Carbon_Emission_tCO2e_TARGET'].rolling(window=30).mean()
rolling_var = df_cleaned['Carbon_Emission_tCO2e_TARGET'].rolling(window=30).var()

plt.figure(figsize=(15, 5))
plt.plot(df_cleaned['Date'], df_cleaned['Carbon_Emission_tCO2e_TARGET'], label='Original Emissions', alpha=0.3)
plt.plot(df_cleaned['Date'], rolling_mean, label='Rolling Mean (30 Days)', color='red', linewidth=2)
plt.title('Rolling Mean of Carbon Emissions')
plt.legend()
plt.show()

plt.figure(figsize=(15, 5))
plt.plot(df_cleaned['Date'], rolling_var, label='Rolling Variance (30 Days)', color='green', linewidth=2)
plt.title('Rolling Variance of Carbon Emissions')
plt.legend()
plt.show()

# 4. تفكيك السلسلة الزمنية (Decompose Time Series: Trend & Seasonality)
daily_data = df_cleaned.set_index('Date')['Carbon_Emission_tCO2e_TARGET'].resample('D').mean().ffill()

decomposition = seasonal_decompose(daily_data, model='additive', period=30)
fig = decomposition.plot()
fig.set_size_inches(15, 10)
plt.show()
```

```python
import numpy as np
from scipy.stats import skew

print("--- Answers to Advanced Questions ---\n")

# Q1: Which emission sources have highest variability?
sources = ['Renewable_Energy_Consumption_kWh', 'NonRenewable_Energy_Consumption_kWh', 'Total_Energy_Consumption_kWh']
variances = df_cleaned[sources].var()
highest_var_source = variances.idxmax()
print(f"Q1: Which emission sources have highest variability?")
print(f"Answer: The source with the highest variability is '{highest_var_source}' with a variance of {variances.max():.2f}\n")

# Q2: Are emissions normally distributed?
emissions_skew = skew(df_cleaned['Carbon_Emission_tCO2e_TARGET'])
print(f"Q2: Are emissions normally distributed?")
if abs(emissions_skew) < 0.5:
    print(f"Answer: Yes, they are approximately normally distributed. (Skewness = {emissions_skew:.2f}, which is close to 0)\n")
else:
    print(f"Answer: No, they are NOT perfectly normally distributed. (Skewness = {emissions_skew:.2f})\n")

# Q3: Which variables are strongly correlated?
numeric_cols = df_cleaned.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_cols.corr().abs()

np.fill_diagonal(corr_matrix.values, 0)
max_corr = corr_matrix.unstack().sort_values(ascending=False)
var1, var2 = max_corr.index[0]
print(f"Q3: Which variables are strongly correlated?")
print(f"Answer: The strongest correlation is between '{var1}' and '{var2}' (Correlation coefficient = {max_corr.iloc[0]:.2f})\n")

# Q4: Is there a trend in emissions?
x = np.arange(len(df_cleaned))
y = df_cleaned['Carbon_Emission_tCO2e_TARGET'].values
slope, _ = np.polyfit(x, y, 1)
print(f"Q4: Is there a trend in emissions?")
if slope > 0.01:
    print("Answer: Yes, there is a slight UPWARD trend in emissions over time.\n")
elif slope < -0.01:
    print("Answer: Yes, there is a slight DOWNWARD trend in emissions over time.\n")
else:
    print("Answer: No significant long-term trend exists; emissions are relatively stable over time.\n")

# Q5: Do seasonal patterns exist?
print(f"Q5: Do seasonal patterns exist?")
print("Answer: Based on the Seasonal Decompose plot from the previous step, if the 'Seasonal' subplot shows a repeating, wave-like pattern, then yes, seasonal patterns exist in the data.")
```
